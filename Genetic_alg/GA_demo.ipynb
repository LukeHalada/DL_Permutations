```python
from generate_tsp import TSP
from TSP_GA import GA_TSP
import numpy as np
```

### Generate a TSP problem

```python
#tsp parameters
tsp_params = {
    'cities_count': 20,
    'min': 0,
    'max': 10,
    'start': 1,
    'seed': 0
}
```

```python
tsp = TSP(**tsp_params)
tsp.cities
```

### Visualize a sample solution

```python
random_solution = tsp.generate_solutions(1)
print(f"Solution: {random_solution}\n Total distance: {tsp.total_distance(random_solution[0])}")
```

```python
tsp.plot_solution(random_solution[0], figsize=(5,5))
```

### Use genetic algorithm to find a good solution

```python
#hyperparameters
ga_params = {
'cities': tsp.cities,
'population_size': 50,
'mutation_rate': 0.1,
'cross_rate': 0.1,
'tournament_size': 15,
'fitness': tsp.total_distance
}
```

```python
ga = GA_TSP(**ga_params)
population, history = ga.fit(epochs=30,
                            starting_population=tsp.generate_solutions(50),)
```

```python
tsp.plot_solution(min(population, key=lambda x: tsp.total_distance(x)), figsize=(5,5))
```
